# Encapsulation (Enkapsulyatsiya)

```python
class BankHisobi:
    def __init__(self, balance):
        self.__balance = balance  

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount

hisob = BankHisobi(10000)
print(hisob.get_balance())  
hisob.deposit(35000)
print(hisob.get_balance())  
```

# Inheritance (Meros olish )

```python
class Hayvon:
    def speak(self):
        print("Hayvon Gapirishi")

class Kuchug(Hayvon):  
    def speak(self):
        print("Woof! Woof!")

g = Kuchug()
g.speak()  
```

# Polymorphism (Ko`p Shakllilik) 

```python
class Cat:
    def sound(self):
        print("Meow! Meow!")

class Dog:
    def sound(self):
        print("Woof! Woof!")


animals = [Cat(), Dog()]
for animal in animals:
    animal.sound()
```

```python
class Talaba:
    def __init__(self, name, age):
        self.name = name
        self.age = age 

    def introduce(self):
        print(f"Mening ismim {self.name} va Yoshim {self.age} da")

a = Talaba("Rasulbek", 32)
a.introduce()
```

# Misol tortburchak ni topish 

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

r = Rectangle(8, 15)
r.area()  
```

# Misol Uchburchakni topish 

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return math.pi * self.radius ** 2

c = Circle(5)
print(c.area())  
```

# Misol Bankdagi hisob  haqida xisobot

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Hisobda pul yo`q")

acc = BankAccount("Rasulbek")
acc.deposit(15000)
acc.withdraw(5000)
print(acc.balance)  
```
